```python
#2.2
def square_elements(numbers):
    squared_numbers = [x**2 for x in numbers]
    return squared_numbers

Rats = list(range(21))  
squared_numbers = square_elements(Rats)

print(squared_numbers)
```

```python
#2.3
def get_first_15_elements(numbers):
    return numbers[:15]


Rats = list(range(21))  
squared_numbers = [x**2 for x in Rats]  
first_15_squared_numbers = get_first_15_elements(squared_numbers)


print(first_15_squared_numbers)
```

```python
#2.4
def get_every_5th_element(numbers):
    return numbers[::5]

numbers_0_to_20 = list(range(21))  
squared_numbers = [x**2 for x in numbers_0_to_20]  
every_5th_squared_number = get_every_5th_element(squared_numbers)


print(every_5th_squared_number)
```

```python
#2.5
def get_reverse_every_3rd_element(numbers):
    last_30_elements = numbers[-30:]
    
    return last_30_elements[::-3]

numbers_0_to_20 = list(range(21))  
squared_numbers = [x**2 for x in numbers_0_to_20]
reverse_every_3rd_squared_number = get_reverse_every_3rd_element(squared_numbers)

print(reverse_every_3rd_squared_number)
```

```python
#3.1
def create_5x5_matrix():
    matrix = []
    num = 1
    
    for i in range(5): 
        row = [] 
        for j in range(5):   
            row.append(num)  
            num += 1  
        matrix.append(row)  
    
    return matrix

matrix_5x5 = create_5x5_matrix()

for row in matrix_5x5:
    print(row)
```

```python
#3.2
def replace_multiples_of_3(matrix):
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):  
            if matrix[i][j] % 3 == 0:  
                matrix[i][j] = "?"
    
    return matrix

matrix_5x5 = create_5x5_matrix() 


modified_matrix = replace_multiples_of_3(matrix_5x5)

for row in modified_matrix:
    print(row)
```

```python
#3.3
def sum_non_question_marks(matrix):    total_sum = 0
    for row in matrix:
        
        for element in row: 
            
            if element != "?":

                total_sum += element 
    
    return total_sum

modified_matrix = replace_multiples_of_3(create_5x5_matrix())  

result = sum_non_question_marks(modified_matrix)

print("Sum of non-'?' elements:", result)
```
