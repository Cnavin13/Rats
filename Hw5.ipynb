```python
def find_rows_with_all_odd(arr):
    odd_rows = []
    
    for i, row in enumerate(arr):
        if np.all(row % 2 != 0): 
            odd_rows.append(i)
    
    return odd_rows

arr = np.array([[1, 2, 3], [5, 7, 9], [2, 4, 6], [7, 7, 7]])
rows_with_all_odd = find_rows_with_all_odd(arr)

print("Rows with all odd numbers:", rows_with_all_odd)
```

```python
#2.1
def create_zero_matrix():
    return np.zeros((8, 8), dtype=int)

zero_matrix = create_zero_matrix()

print(zero_matrix)
```

```python
#2.2
def create_alternating_matrix():
    matrix = np.zeros((8, 8), dtype=int)
    
    for i in range(0, 8, 2):  
        matrix[i] = [1 if j % 2 == 0 else 0 for j in range(8)]  
    
    return matrix

alternating_matrix = create_alternating_matrix()

print(alternating_matrix)
```

```python
#2.3
def create_checkerboard_matrix():
    matrix = np.zeros((8, 8), dtype=int)
    
    for i in range(0, 8, 2):  
        matrix[i] = [1 if j % 2 == 0 else 0 for j in range(8)]  
    
    for i in range(1, 8, 2):  
        matrix[i] = [0 if j % 2 == 0 else 1 for j in range(8)]  
    
    return matrix

checkerboard_matrix = create_checkerboard_matrix()

print(checkerboard_matrix)
```

```python
#2.4
def create_checkerboard_matrix():
    matrix = np.zeros((8, 8), dtype=int)
    
    
    for i in range(0, 8, 2):
        matrix[i] = [0 if j % 2 == 0 else 1 for j in range(8)]  
    
    for i in range(1, 8, 2):
        matrix[i] = [1 if j % 2 == 0 else 0 for j in range(8)]  
    
    return matrix

checkerboard_matrix = create_checkerboard_matrix()

print(checkerboard_matrix)
```

```python
def expand_string_with_spaces(s, num_spaces):
   
    space = ' ' * num_spaces
    
    expanded_string = space.join(s)
    
    return expanded_string

universe = np.array(['galaxy', 'clusters'])

expanded_universe = np.array([expand_string_with_spaces(word, 2) for word in universe])

print(expanded_universe)
```

```python
def second_largest_in_columns(arr):
    
    second_largest_values = []
     
    for col in arr.T:  
        sorted_col = np.sort(col)  
        second_largest_values.append(sorted_col[-2])  
        
    return np.array(second_largest_values)

np.random.seed(42)  
arr = np.random.randint(1, 100, (5, 4))  

print("Original Array:")
print(arr)


result = second_largest_in_columns(arr)

print("Second-largest values in each column:")
print(result)
```
