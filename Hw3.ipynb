## 1) Oski Stole your power

```python
def power(x, y):
    
    if y == 0:
        return 1  
    elif y > 0:
        result = 1
        for _ in range(y):
            result *= x
        return result
    else:  
        result = 1
        for _ in range(-y):
            result *= x
        return 1 / result


print(power(2, 3))  
```

## 2) What should I wear?

```python
def temperatureRange(readings):
    
    if not readings:
        raise ValueError("The list is empty. Please provide a valid list.")
    
    minimum = float('inf')
    maximum = float('-inf')
    
    for value in readings:
        if value < minimum:
            minimum = value
        if value > maximum:
            maximum = value
    
    return (minimum, maximum)

readings = [15, 14, 17, 20, 23, 28, 20]
print(temperatureRange(readings))  
```

## 3) Check if its the weekend

```python
def isWeekend(day):
    
    if day < 1 or day > 7:
        raise ValueError("Invalid day. Please provide a number between 1 and 7.")
    
    return day == 6 or day == 7


day = 6  
print(isWeekend(day))  

day = 3  
print(isWeekend(day)) 
```

```python
def fuel_efficiency(distance, fuel):
   
    if fuel <= 0:
        raise ValueError("Fuel consumed must be greater than 0.")
    if distance < 0:
        raise ValueError("Distance traveled cannot be negative.")
    
    return distance / fuel


distance = 70  
fuel = 21.5  
print(fuel_efficiency(distance, fuel))  
```

```python
def move_last_digit_to_front(n):
   
    is_negative = n < 0
    n = abs(n)  
    
    last_digit = n % 10 
    remaining_number = n // 10  
    
   
    if remaining_number == 0:
        return -last_digit if is_negative else last_digit  
    
    multiplier = 1
    temp = remaining_number
    while temp > 0:
        temp //= 10
        multiplier *= 10
    
    result = last_digit * multiplier + remaining_number
    return -result if is_negative else result

print(move_last_digit_to_front(12345))  
print(move_last_digit_to_front(-9876))  
print(move_last_digit_to_front(5))  
```

```python
def find_min(nums):
    
    if not nums:
        raise ValueError("The list is empty.")
    
    min_val = nums[0]
    for num in nums:
        if num < min_val:
            min_val = num
    return min_val

def find_max(nums):
    
    if not nums:
        raise ValueError("The list is empty.")
    
    max_val = nums[0]
    for num in nums:
        if num > max_val:
            max_val = num
    return max_val


nums = [2024, 98, 131, 2, 3, 72]
print("Minimum value:", find_min(nums))
print("Maximum value:", find_max(nums))  
```

```python
def find_min(nums):
   
    if not nums:
        raise ValueError("The list is empty.")
    
    min_val = nums[0]
    i = 1
    while i < len(nums):
        if nums[i] < min_val:
            min_val = nums[i]
        i += 1
    return min_val

def find_max(nums):
   
    if not nums:
        raise ValueError("The list is empty.")
    
    max_val = nums[0]
    i = 1
    while i < len(nums):
        if nums[i] > max_val:
            max_val = nums[i]
        i += 1
    return max_val


nums = [2024, 98, 131, 2, 3, 72]
print("Minimum value:", find_min(nums))
print("Maximum value:", find_max(nums))  
```

```python
def count_vowels_and_consonants(text):
    
    vowels = "aeiouAEIOU"
    vowel_count = 0
    consonant_count = 0
    
    for char in text:
        if char.isalpha():  
            if char in vowels:
                vowel_count += 1
            else:
                consonant_count += 1
    
    return vowel_count, consonant_count


text = "UC Berkeley, founded in 1868!"
result = count_vowels_and_consonants(text)
print("Vowels:", result[0])     
print("Consonants:", result[1])  
```

```python
def sum_of_digits(num):
   
    num = abs(num)  
    digit_sum = 0
    
    while num > 0:
        digit_sum += num % 10
        num //= 10  
    
    return digit_sum

num = 2468
print(sum_of_digits(num))  
```
